```python
def convert_ssf_to_conll(input_ssf):
    conll_lines = []
    sentences = input_ssf.strip().split('\n\n')
    for sentence in sentences:
        lines = sentence.strip().split('\n')
        for line in lines:
            if "((\t" in line:
                continue
            if "</Sentence" in line:
              conll_lines.append("")
            tokens = line.strip().split('\t')
            if len(tokens) > 1:
                word = tokens[1]
                pos = tokens[2]
                conll_lines.append(f"{word}\t{pos}")
    return '\n'.join(conll_lines)

for i in train_files:
  fname = '/content/Data/Training/' + i
  f = open(fname, 'r')
  ssf_input = f.read()
  f.close()
  conll_output = convert_ssf_to_conll(ssf_input)
  out_file = '/content/Data/CoNLL/Training/' + i
  out = open(out_file, 'w')
  out.write(conll_output)
  out.close()
for i in test_files:
  fname = '/content/Data/Testing/' + i
  f = open(fname, 'r')
  ssf_input = f.read()
  f.close()
  conll_output = convert_ssf_to_conll(ssf_input)
  out_file = '/content/Data/CoNLL/Testing/' + i
  out = open(out_file, 'w')
  out.write(conll_output)
  out.close()
for i in validate_files:
  fname = '/content/Data/Development/' + i
  f = open(fname, 'r')
  ssf_input = f.read()
  f.close()
  conll_output = convert_ssf_to_conll(ssf_input)
  out_file = '/content/Data/CoNLL/Development/' + i
  out = open(out_file, 'w')
  out.write(conll_output)
  out.close()
```

Define Emission Probability and Transition Probability

```python
# compute Emission Probability
def word_given_tag(word, tag, train_bag = train_tagged_words):
    tag_list = [pair for pair in train_bag if pair[1]==tag]
    count_tag = len(tag_list)#total number of times the passed tag occurred in train_bag
    w_given_tag_list = [pair[0] for pair in tag_list if pair[0]==word]
#now calculate the total number of times the passed word occurred as the passed tag.
    count_w_given_tag = len(w_given_tag_list)


    return (count_w_given_tag, count_tag)

# compute  Transition Probability
def t2_given_t1(t2, t1, train_bag = train_tagged_words):
    tags = [pair[1] for pair in train_bag]
    count_t1 = len([t for t in tags if t==t1])
    count_t2_t1 = 0
    for index in range(len(tags)-1):
        if tags[index]==t1 and tags[index+1] == t2:
            count_t2_t1 += 1
    return (count_t2_t1, count_t1)
```

```python
# creating t x t transition matrix of tags, t= no of tags
# Matrix(i, j) represents P(jth tag after the ith tag)

tags_matrix = np.zeros((len(tags), len(tags)), dtype='float32')
for i, t1 in enumerate(list(tags)):
    for j, t2 in enumerate(list(tags)):
        tags_matrix[i, j] = t2_given_t1(t2, t1)[0]/t2_given_t1(t2, t1)[1]
```

```python
print(tags_matrix)
```

```python
# convert the matrix to a dataframe for better readability
#the table is same as the transition table shown in section 3 of article
tags_df = pd.DataFrame(tags_matrix, columns = list(tags), index=list(tags))
print(tags_df)
```

Define Viterbi Algorithm

```python
def Viterbi(words, train_bag = train_tagged_words):
    state = []
    T = list(set([pair[1] for pair in train_bag]))

    for key, word in enumerate(words):
        #initialise list of probability column for a given observation
        p = []
        for tag in T:
            if key == 0:
                transition_p = 0
            else:
                transition_p = tags_df.loc[state[-1], tag]

            # compute emission and state probabilities
            emission_p = word_given_tag(words[key], tag)[0]/word_given_tag(words[key], tag)[1]
            state_probability = emission_p * transition_p
            p.append(state_probability)

        pmax = max(p)
        # getting state for which probability is maximum
        state_max = T[p.index(pmax)]
        state.append(state_max)
    return list(zip(words, state))
```

```python
#Class purpose CL-2026 March 13

#import numpy as np

# 1. Define Model Parameters (from your slide tables)
# States: N (Noun), M (Model), V (Verb)
states = ['N', 'M', 'V']

# Transition Probabilities P(St | St-1) [cite: 322]
# Rows: <S>, N, M, V | Columns: N, M, V, <E>
transitions = {
    '<S>': {'N': 3/4, 'M': 1/4, 'V': 0},
    'N':   {'N': 1/9, 'M': 3/9, 'V': 1/9, '<E>': 4/9},
    'M':   {'N': 1/4, 'M': 0,   'V': 3/4, '<E>': 0},
    'V':   {'N': 1,   'M': 0,   'V': 0,   '<E>': 0}
}

# Emission Probabilities P(Ot | St) [cite: 250]
emissions = {
    'N': {'Will': 1/9, 'can': 0,   'spot': 2/9, 'Mary': 4/9},
    'M': {'Will': 3/4, 'can': 1/4, 'spot': 0,   'Mary': 0},
    'V': {'Will': 0,   'can': 0,   'spot': 1/4, 'Mary': 0}
}

def calculate_path_probability(sentence, tags):
    """Calculates P(Words, Tags) = P(Tags) * P(Words|Tags)"""
    prob = 1.0

    # Start transition: P(Tag1 | <S>)
    prob *= transitions['<S>'][tags[0]]

    for i in range(len(sentence)):
        # Emission: P(Word_i | Tag_i)
        prob *= emissions[tags[i]].get(sentence[i], 0)

        # Transition: P(Tag_i+1 | Tag_i)
        if i < len(sentence) - 1:
            prob *= transitions[tags[i]][tags[i+1]]
        else:
            # End transition: P(<E> | Last Tag)
            prob *= transitions[tags[i]]['<E>']

    return prob

# 2. Test the two paths from your slide 38 [cite: 441, 445]
sentence = ["Will", "can", "spot", "Mary"]

path1 = ['N', 'M', 'N', 'N']
path2 = ['N', 'M', 'V', 'N']

prob1 = calculate_path_probability(sentence, path1)
prob2 = calculate_path_probability(sentence, path2)

print(f"Path 1 (N-M-N-N) Prob: {prob1:.10f}") # [cite: 444]
print(f"Path 2 (N-M-V-N) Prob: {prob2:.10f}") # [cite: 447]
print(f"\nWinner: {'Path 2' if prob2 > prob1 else 'Path 1'}")
```
